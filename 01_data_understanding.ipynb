```python
year_coverage = (
    df.groupby("geo")
    .agg(
        first_year=("TIME_PERIOD", "min"),
        last_year=("TIME_PERIOD", "max"),
        number_of_years=("TIME_PERIOD", "nunique")
    )
    .sort_values("number_of_years")
)
year_coverage
```

```python
years_per_country=(
    df.groupby("geo")["TIME_PERIOD"]
    .apply(lambda years: sorted(years.unique()))
)
years_per_country
```

```python
flag_summary=(
    df[df["OBS_FLAG"].notna()]
    .groupby(["OBS_FLAG", "geo"])
    .size()
    .reset_index(name="number_of_rows")
)
flag_summary
```

```python
df.loc[
    df["OBS_FLAG"] == "d",
    ["geo", "TIME_PERIOD", "sex", "OBS_VALUE", "OBS_FLAG"]
].sort_values(["geo", "TIME_PERIOD", "sex"])
```

```python
def find_missing_years(years):
    existing_years = set(years)
    expected_years = set(range(min(years), max(years) + 1))

    return sorted(expected_years - existing_years)


missing_years = (
    df.groupby("geo")["TIME_PERIOD"]
      .apply(find_missing_years)
)

missing_years[missing_years.apply(len) > 0]
```
